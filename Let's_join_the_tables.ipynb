```python
import numpy as np
import pandas as pd
import sqlite3
```

```python
database='/content/basketball.sqlite'
conn=sqlite3.connect(database)
print('Opened data successfully')
```

```python
tables=pd.read_sql("""SELECT *
                        FROM sqlite_master
                        WHERE type='table';""",conn)
tables
```

```python
player=pd.read_sql("""SELECT *
                        FROM Player;""",conn)
player
```

```python
salary=pd.read_sql("""SELECT *
                        FROM Player_Salary;""",conn)
salary
```

```python
join=pd.read_sql("""SELECT DISTINCT ps.nameTeam, ps.namePlayer, ps.value, p.is_active
                    FROM Player p
                    JOIN Player_Salary ps
                    ON p.full_name = ps.namePlayer;""",conn)
join
```
