```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
xpoints=np.array([0,6])
ypoints=np.array([0,250])
plt.plot(xpoints,ypoints)
plt.show()
```

```python
xpoints=np.array([1,8])
ypoints=np.array([3,10])
plt.plot(xpoints,ypoints,'o')
plt.show()
```

```python
xpoints=np.array([1,2,6,8])
ypoints=np.array([3,8,1,10])
plt.plot(xpoints,ypoints)
plt.show()
```

```python
ypoints=np.array([3,8,1,10])
plt.plot(ypoints,marker='o')
plt.show()
```

```python
ypoints=np.array([3,8,1,10])
plt.plot(ypoints,marker='o',ms=20,mec='r')
#ms=marker size,mec=marker edge colour
plt.show()
```

```python
ypoints=np.array([3,8,1,10])
plt.plot(ypoints,linewidth='20.5')
plt.show()
```

```python
y1=np.array([3,8,1,10])
y2=np.array([6,2,7,11])
plt.plot(y1)
plt.plot(y2)
plt.show()
```

```python
x=np.array([80,85,90,100,105,110,115,120,125,130])
y=np.array([200,250,260,270,280,290,300,310,320,330])
plt.plot(x,y)
plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")
plt.show()
```

```python
students_names=["Ariana","Tathoi","Sayeema","Zawata","Mahek"]
students_marks=[80,76,82,78,72]
marks_perc=[]
for x in students_marks:
    res=(x/100)*100
    marks_perc.append(res)
print(marks_perc)
```

```python
#line chart
def marks_line_chart():
  plt.plot(students_names,students_marks)
  plt.title("Students Marks Graph")
  plt.xlabel("Students Name")
  plt.ylabel("Students Marks")
  plt.show()
marks_line_chart()
```

```python
#bar chart
def percentage_bar_chart():
  plt.bar(students_names,marks_perc)
  plt.title("Students Percentage Graph")
  plt.xlabel("Students Name")
  plt.ylabel("Students Percentage")
  plt.show()
percentage_bar_chart()
```
