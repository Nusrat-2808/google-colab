```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('/content/Company Sales Data.csv')
```

```python
plt.plot(data['month_number'], data['total_profit'],
         linestyle='dotted', marker='o', color='red',
         linewidth=3, markerfacecolor='black')
plt.xlabel('Month Number')
plt.ylabel('Profit')
plt.title('Company Profit per Month')
plt.show()
```

```python
for column in data.columns[1:-1]:
  # skipping month_number and total_profit
  plt.plot(data['month_number'], data[column],
           marker='o', linewidth=3, label=column)
plt.xlabel('Month Number')
plt.ylabel('Sales Units')
plt.title('Sales Data of All Products per Month')
plt.legend()
plt.show()
```

```python
plt.bar(data['month_number'], data['facecream'],
        label='Face Cream', width=0.4, align='center')
plt.bar(data['month_number'], data['facewash'],
        label='Face Wash', width=0.4, align='edge')
plt.xlabel('Month Number')
plt.ylabel('Sales Units')
plt.title('Face Cream vs Face Wash Sales per Month')
plt.legend()
plt.show()
```
