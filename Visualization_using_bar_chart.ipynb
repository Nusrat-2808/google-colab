```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
july_data = {'Day':['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
             'New Birth':[12,15,11,9,1,9,21]}
july_df = pd.DataFrame(july_data)
```

```python
plt.bar(july_df['Day'], july_df['New Birth'],
         linewidth=3,color='C5',
        edgecolor="firebrick",label='July')
plt.title("New Birth Data-July")
plt.xlabel("Day of July")
plt.ylabel("New Birth")
plt.legend()
plt.show()
```

```python
aug_data = {'Day':['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
            'New Birth':[17,5,2,11,1,8,29]}
aug_df = pd.DataFrame(aug_data)
```

```python
plt.bar(aug_df['Day'], aug_df['New Birth'],
         linewidth=3,color='C8',
         edgecolor="C9",label='August')
plt.title("New Birth Data-August")
plt.xlabel("Day of August")
plt.ylabel("New Birth")
plt.legend()
plt.show()
```

```python
y1=[12,15,11,9,1,9,21]
y2=[17,5,2,11,1,8,29]
barwidth=0.25
br1=np.arange(len(y1))
br2=[x+barwidth for x in br1]
plt.bar(br1,y1,width=barwidth,color='firebrick',label='July')
plt.bar(br2,y2,width=barwidth,linewidth=3,color='C9',label='August')
plt.title("Comparison of new birth in 2 months")
plt.xlabel("Day of month")
plt.xticks([x+barwidth for x in br1],['Monday','Tuesday','Wednesday',
                                          'Thursday','Friday','Saturday',
                                          'Sunday'],rotation=90)
plt.ylabel("New Birth")
plt.legend()
plt.show()
```
