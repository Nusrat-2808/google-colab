```python
import random
outcomes=[1,2,3,4,5,6]
```

```python
Total_outcomes=len(outcomes)
Favourable_outcome=outcomes.count(6)
probability_6=Favourable_outcome/Total_outcomes
print(probability_6)
```

```python
def roll_a_dice(List):
  result=random.choice(List)
  if result==6:
    print("The Game can be started!")
  else:
    print("Roll again.Best of luck!")
roll_a_dice(outcomes)
```
