```python
# Basket composition
red = 1
blue = 6
white = 3
total = red + blue + white

# Probability that second ball is white given first is white (with replacement)
prob_a = white / total
prob_b = white / total
prob_a_and_b = prob_a * prob_b
print("Probability that the second ball is white given the first is white(with replacement):", prob_a_and_b)

#Probability that second ball is white given first is white(without replacement)
prob_a=white/total
prob_b=(white-1)/(total-1)
prob_a_and_b=prob_a*prob_b
print("Probability that the second ball is white given the first is white(without replacement):", prob_a_and_b)
```
