```python
import scipy.stats as stats
```

```python
#Probability of observing 12 or more days of rain in next 30 days
#Lambda=10
prob1= 1-stats.poisson.cdf(11,10)
print(prob1)
```

```python
#Probability of observing between 12 and 18 dsays of raining in next 30 days
#Lambda=10
#Method 1
prob2=stats.poisson.pmf(12,10) + stats.poisson.pmf(13,10) + stats.poisson.pmf(14,10) + stats.poisson.pmf(15,10) + stats.poisson.pmf(16,10) + stats.poisson.pmf(17,10) + stats.poisson.pmf(18,10)
print("Probability of raining 12-18 days")
print(prob2)
```

```python
#Probability of observing between 12 and 18 dsays of raining in next 30 days
#Lambda=10
#Method 2
prob2=stats.poisson.cdf(18,10) - stats.poisson.cdf(11,10)
print("Probability of raining 12-18 days")
print(prob2)
```
