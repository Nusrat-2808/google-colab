```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
df
```

```python
sns.barplot(x='day',y='total_bill',data=df)
```

```python
sns.boxplot(x='day',y='total_bill',data=df)
```

```python
sns.countplot(x='smoker',data=df)
```

```python
sns.scatterplot(y='total_bill',x='tip',data=df)
```

```python
sns.stripplot(x='tip',y='total_bill',data=df,palette="plasma")
```

```python
sns.swarmplot(x='day',y='total_bill',data=df,palette="plasma")
```

```python
sns.pointplot(x='smoker',y='total_bill',data=df,palette="plasma")
```

```python
sns.lmplot(x='total_bill',y='tip',data=df,palette="plasma")
```

```python
sns.pairplot(df)
```
