```python
Quantile10=np.quantile(df["Price"],0.10)
print("Value dividing 10% of Price is-",Quantile10)
```

```python
Price_Q1=np.quantile(df["Price"],0.25)
print("First quantile of Price is-",Price_Q1)
Price_Q2=np.quantile(df["Price"],0.50)
print("Second quantile of Price is-",Price_Q2)
Price_Q3=np.quantile(df["Price"],0.75)
print("Third quantile of Price is-",Price_Q3)
```

```python
IQR_Price=Price_Q3-Price_Q1
print("Inter Quantile Range of Price is-",IQR_Price)
```

```python
User_rating_Q1=np.quantile(df["User Rating"],0.25)
print("First quantile of User Rating is-",User_rating_Q1)
User_rating_Q2=np.quantile(df["User Rating"],0.50)
print("Second quantile of User Rating is-",User_rating_Q2)
User_rating_Q3=np.quantile(df["User Rating"],0.75)
print("Third quantile of User Rating is-",User_rating_Q3)
```

```python
IQR_User_rating=User_rating_Q3-User_rating_Q1
print("Inter Quantile Range of User Rating is-",IQR_User_rating)
```
