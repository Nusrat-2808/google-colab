# ACTIVITY 1

```python
import numpy as np
```

```python
a=np.arange(9,dtype=np.float64).reshape(3,3)
print("FIRST ARRAY:")
print(a)
print('\n')
```

```python
b=np.array([10,10,10])
print("SECOND ARRAY:")
print(b)
print('\n')
```

```python
print("ADD THE TWO ARRAYS:")
print(np.add(a,b))
print('\n')
```

```python
print("SUBTRACT THE TWO ARRAYS:")
print(np.subtract(a,b))
print('\n')
```

```python
print("MULTIPLY THE TWO ARRAYS:")
print(np.multiply(a,b))
print('\n')
```

```python
print("DIVIDE THE TWO ARRAYS:")
print(np.divide(a,b))
print('\n')
```

# ACTIVITY 2

```python
import numpy as np
```

```python
#DEFINE THE STRUCTURED DATA
data_type=[('name','U15'),('class',int),('height',float)]
#define student details
students_details=[('James',5,48.5),
                  ('Nail',6,52.5),
                  ('Paul',5,42.10),
                  ('Pit',5,40.11)]
```

```python
#create a structured numpy array
students=np.array(students_details,dtype=data_type)
#print the original array
print("ORIGINAL ARRAY:")
print(students)
```

```python
#sort by height
print("\n Sort by height:")
print(np.sort(students,order='height'))
```

```python
#sort by class
print("\n Sort by class:")
print(np.sort(students,order='class'))
```

```python
#sort by height
print("Students with more than height 50:")
student2=students[students['height']>50]
print(student2)
```
