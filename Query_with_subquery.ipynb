```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
import sqlite3
database='/content/database.sqlite'
conn=sqlite3.connect(database)
tables=pd.read_sql("""SELECT * FROM
                    sqlite_master
                    WHERE type='table';""",conn)
tables
```

```python
#Aliasing
match_details=pd.read_sql("""SELECT Season_Id,Match_Id,
                            v.Venue_Name,c.City_Name,t.Team_Name AS Winner
                            FROM Match
                            INNER JOIN Venue AS v ON match.Venue_Id==v.Venue_Id
                            INNER JOIN City AS c ON v.City_Id==c.City_Id
                            INNER JOIN Team AS t ON match.Match_Winner==t.Team_Id;""",conn)
match_details
```

```python
import pandas as pd
import numpy as np
```

```python
import sqlite3
database='/content/database (1).sqlite'
conn=sqlite3.connect(database)
tables=pd.read_sql("""SELECT * FROM
                    sqlite_master
                    WHERE type='table';""",conn)
tables
```

```python
team=pd.read_sql("""SELECT * FROM Team""",conn)
team
```

```python
season=pd.read_sql("""SELECT * FROM Season""",conn)
season
```

```python
csk_matches_2015=pd.read_sql("""SELECT Match_Id,Team_2 AS Away_Team,Toss_Winner,Match_Winner
                              FROM Match
                              WHERE Team_1=
                              (SELECT Team_1
                              FROM Match
                              WHERE Team_1==3 AND Season_Id==8)""",conn)
print("Matches played by chennai super kings in year 2015")
csk_matches_2015
```

```python
csk_wins=pd.read_sql("""SELECT *
                        FROM Match
                        WHERE Match_Winner==3 AND Season_Id==8""",conn)
print("Matches won by CSK as Home Team in year 2015")
csk_wins
```

```python
match_runs=pd.read_sql("""SELECT Match_Id,Runs_Scored as Total_Runs,Innings_No
                          FROM Batsman_Scored
                          WHERE Total_Runs>5 AND Match_Id IN
                          (SELECT Match_Id
                          FROM Match
                          WHERE Season_Id==8)""",conn)
print("Matches with scored runs greater than 5 in year 2015")
match_runs
```

```python
avg_runs=pd.read_sql("""SELECT Match_Id,Runs_Scored AS Total_Runs,Innings_No
                        FROM Batsman_Scored
                        WHERE Innings_No==1 AND Runs_Scored>
                        (SELECT AVG(Runs_Scored)
                        FROM Batsman_Scored)""",conn)
print("Matches with scored runs greater than average scored runs")
avg_runs
```
