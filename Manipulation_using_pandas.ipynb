```python
import pandas as pd
import numpy as np
```

```python
data={
    "Id":[1,2,3,4],
    "Name":['Pankaj','Meghna','David','Lisa'],
    "Role":['CEO',np.nan,np.nan,np.nan],
    "Salary":[100,200,np.nan,np.nan]
}
```

```python
df=pd.DataFrame(data)
```

```python
print(df)
```

```python
print(df.head(2))
```

```python
print(df.tail(2))
```

```python
print(df.info())
```

```python
print("Total null values:", df.isnull().sum())
```

```python
df_rows_clean = df.dropna()
print(df_rows_clean)

df_cols_clean = df.dropna(axis=1)
print(df_cols_clean)
```

```python
df["Salary"].fillna(300)
df["Role"].fillna("CEO")
print(df)
```
