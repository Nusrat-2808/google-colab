```python
data=pd.read_csv("Titanic Dataset.csv")
data.head(5)
```

```python
###Passengers belonging from which gender survives the most
sns.countplot(x='Gender',hue='Survived',data=data)
#
```

```python
###Passengers belonging from which PClass survived the most and the least
sns.countplot(x='Pclass',hue='Survived',data=data)
```

```python
###Highest number of passengers belong to which Age
sns.distplot(data['Age'],kde=False,bins=40)
```

```python
###Highest number of passengers belong to which gender
sns.countplot(x='Gender',data=data)
```

```python
###Is SibSp correlated/associated with survived feature
sns.countplot(x='Survived',hue='SibSp',data=data,palette='mako')
```

```python
###Is Parch correlated/associated with survived feature
sns.countplot(x='Survived',hue='Parch',data=data,palette='mako')
```

```python
###Is the feature fare having normal distributtion /spread of data
sns.distplot(data['Fare'])
plt.show()
```

```python
###Check the age group of majority of people belonging to Pclass=1
sns.boxplot(x='Pclass',y='Age',data=data,palette='winter')
```

```python
###Check the correlation of all the features with target variable survived
sns.heatmap(data.corr(numeric_only=True))
```

```python
data=pd.read_csv("Iris Dataset.csv")
data.head(5)
```

```python
data.isnull().sum()
```

```python
##Statistical information of features
data.describe()
```

```python
##Create boxplot of all features and check if outliers are present
labels=['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for label in labels:
  print('Distribution of', label)
  sns.boxplot(data[label])
  plt.show()
```

```python
##Conclusion
#SepalWidthCm has outliers
```

```python
#Check correlation between all the features
sns.heatmap(data.corr(numeric_only=True))
```

```python
#Check if any feature is skewed or not
labels=['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for label in labels:
  print('Distribution of', label)
  sns.distplot(data[label])
  plt.show()
```

```python
labels=['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
for label in labels:
  print("skewness of ",label)
  print(data[label].skew())
```
