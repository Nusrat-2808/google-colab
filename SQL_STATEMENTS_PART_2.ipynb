```python
result1=pd.read_sql("""SELECT AVG(Win_Margin),Match_Winner
                    FROM Match
                    WHERE Season_Id==9
                    GROUP BY Match_Winner
                    ORDER BY AVG(Win_Margin);""",conn)
result1
```

```python
result2=pd.read_sql("""SELECT COUNT(DISTINCT Venue_Id)
                    FROM Match
                    WHERE Season_id==9;""",conn)
result2
```

```python
result3=pd.read_sql("""SELECT MIN(Win_Margin),MAX(Win_Margin),
                    Avg(Win_Margin),COUNT(DISTINCT(Man_of_the_Match))
                    FROM Match;""",conn)
result3
```
