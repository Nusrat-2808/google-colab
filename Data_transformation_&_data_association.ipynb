```python
data=pd.read_csv("Bestsellers with categories.csv")
data
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
data["Rating_Category"]=pd.cut(data['User Rating'],bins=[0,4,5],labels=['Good','Excellent'])
```

```python
plt.bar(data['Rating_Category'].value_counts().index,data['Rating_Category'].value_counts())
plt.xlabel('Rating Category')
plt.ylabel('Count')
plt.title('Rating Category Distribution')
plt.show()
```

```python
num_cat=['User Rating','Reviews','Price','Year']
for label in num_cat:
  print('Distribution of', label)
  sns.distplot(data[label])
  plt.show()
  print('Skewness -', data[label].skew())
```

```python
sns.boxplot(data = data, x = 'Genre', y = 'User Rating')
plt.show()
```

```python
plt.scatter(x = data['User Rating'], y = data['Price'])
plt.ylabel('Price')
plt.xlabel('User Rating')
plt.show()
```
