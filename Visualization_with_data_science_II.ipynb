```python
import matplotlib.pyplot as plt
```

```python
x=[0,5,10,15,20,25,30]
y1=[10,15,20,20,30,15,0]
y2=[10,12,15,12,20,10,0]
```

```python
plt.plot(x,y1,linestyle='dashed',marker='D')
plt.plot(x,y2,linestyle='dashed',marker='D')
plt.title('velocity-time graph')
plt.xlabel('Velocity m/s')
plt.ylabel('Time s')
plt.xlim(5,25)
plt.ylim(5,25)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.arange(0,10,1)
y1=2*x +1
y2=x+3
```

```python
plt.plot(x,y1,label='y=2x+1',color='green')
plt.plot(x,y2,label='y=x+3',color='blue')
plt.xticks(np.arange(0,10,1))
plt.yticks(np.arange(0,20,1))
plt.grid(True)
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.title('Intersection of Two Linear Equations')
plt.show()
```
