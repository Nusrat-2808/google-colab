```python
df.isnull().sum()
```

```python
var_User_rating=np.var(df['User Rating'])
print("Variance of user rating",var_User_rating)
```

```python
std_User_rating=np.std(df['User Rating'])
print("Standard deviation of user rating",std_User_rating)
```

```python
var_Price=np.var(df['Price'])
print("Variance of price",var_Price)
```

```python
std_Price=np.std(df['Price'])
print("Standard deviation of price",std_Price)
```

```python
df['User Rating'].unique()
```

```python
bins_rating=np.arange(3,5,0.1)
plt.hist(df['User Rating'],bins=bins_rating,edgecolor='black',color='purple')
plt.xlabel('User Rating')
plt.ylabel('Books count')
plt.title('Histogram of User Rating')
```

```python
df['Price'].unique()
```

```python
bins_price=np.arange(0,50,1)
plt.hist(df['Price'],bins=bins_price,edgecolor='black',color='orange')
plt.xlabel('Price')
plt.ylabel('Books count')
plt.title('Histogram of Price')
```
