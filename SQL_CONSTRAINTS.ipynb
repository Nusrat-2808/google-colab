# ACTIVITY 1

```python
import sqlite3
database='CLASS.sqlite'
conn=sqlite3.connect(database)
print("Opened database successfully")
```

```python
conn.execute("""CREATE TABLE CLASS_10
              (SNO INT PRIMARY KEY NOT NULL,
              Roll_No INT NOT NULL,
              NAME TEXT NOT NULL,
              AGE INT DEFAULT(15),
              GENDER TEXT NOT NULL,
              Email_ID TEXT NOT NULL,
              Contact_No REAL NOT NULL);""")
print("Table created successfully")
```

```python
conn.execute("INSERT INTO CLASS_10(SNO,Roll_No,NAME,AGE,GENDER,Email_ID,Contact_No)\
              VALUES(1,1,'ALLEN',14,'MALE','allen@gmail.com',8080900)");
conn.execute("INSERT INTO CLASS_10(SNO,Roll_No,NAME,AGE,GENDER,Email_ID,Contact_No)\
              VALUES(2,2,'BOB',15,'MALE','bob@gmail.com',8080901)");
conn.execute("INSERT INTO CLASS_10(SNO,Roll_No,NAME,AGE,GENDER,Email_ID,Contact_No)\
              VALUES(3,3,'ALISHA',14,'FEMALE','alish@gmail.com',9080900)");
conn.execute("INSERT INTO CLASS_10(SNO,Roll_No,NAME,AGE,GENDER,Email_ID,Contact_No)\
              VALUES(4,4,'JEFF',15,'MALE','jeff@gmail.com',9900900)");
conn.commit()
print("Records created successfully")
```

```python
import pandas as pd
tables=pd.read_sql("""SELECT *
                    FROM sqlite_master
                    WHERE type='table';""",conn)
tables
```

```python
class_10=pd.read_sql("""SELECT * FROM CLASS_10;""",conn)
class_10.head()
```

```python
import pandas as pd
import numpy as np
import sqlite3
database='/content/database.sqlite'
connect=sqlite3.connect(database)
print("Opened data successfully")
df=pd.read_sql("""SELECT * FROM sqlite_master
                WHERE type='table';""",connect)
df
```

```python
player_match=pd.read_sql("""SELECT * FROM Player_Match;""",connect)
player_match.head()
```

```python
null_player_match=pd.read_sql("""SELECT *
                                FROM Player_Match
                                WHERE Team_Id IS NULL;""",connect)
null_player_match
```

```python
toss_dec=pd.read_sql("""SELECT *
                       FROM Match;""",connect)
toss_dec.head()
```
