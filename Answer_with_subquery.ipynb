```python
import sqlite3
database='/content/basketball.sqlite'
conn=sqlite3.connect(database)
```

```python
import pandas as pd
table=pd.read_sql("""SELECT *
                    FROM sqlite_master
                    WHERE type='table';""",conn)
table
```

```python
teams=pd.read_sql("""SELECT *
                    FROM Team;""",conn)
teams
```

```python
team_attributes=pd.read_sql("""SELECT *
                    FROM Team_Attributes;""",conn)
team_attributes
```

```python
result=pd.read_sql("""SELECT *
                    FROM Team_Attributes
                    WHERE ID IN (
                        SELECT id
                        FROM Team
                        WHERE state = 'New York'
                    );""",conn)
result
```
