```python
import random
```

```python
colors=['Blue','Red','Green']
total=len(colors)
```

```python
occur=0
def prob(colors):
  for color in colors:
    if color=='Red':
      occur=+1
    else:
      pass
  probability=occur/total
  print("Probabity of red: ",probability)
```

```python
prob(colors)
```

```python
def check():
  Pick_a_ball=random.choice(colors)
  if Pick_a_ball=='Red':
    print("Red ball was picked.")
  else:
    print(Pick_a_ball)
    print("Red ball was not picked.")
check()
```
