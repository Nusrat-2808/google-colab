```python
import sqlite3
database="/content/basketball.sqlite"
conn=sqlite3.connect(database)
print("Opened database successfully")
```

```python
import pandas as pd
tables=pd.read_sql("""SELECT *
                    FROM sqlite_master
                    WHERE type='table';""",conn)
tables
```

```python
teams=pd.read_sql("""SELECT *
                    FROM Team;""",conn)
teams
```

```python
cities=pd.read_sql("""SELECT * FROM Team WHERE state
                      LIKE 'Ca%';""",conn)
cities
```

```python
salary=pd.read_sql("""SELECT * FROM Player_Salary;""",conn)
salary
```

```python
MIN_MAX_salary=pd.read_sql("""SELECT Min(value) AS MIN_SALARY,
MAX(value) AS MAX_SALARY FROM Player_Salary;""",conn)
MIN_MAX_salary
```
