```python
import random
event1= "We pick out two cards from a standard deck of 52 cards without replacement. Event A is that we pick an Ace on the first draw. Event B is that we pick an Ace on the second draw."
#it is dependent

event2= "Suppose we flip a coin ten times. Event A is that we flip all heads on the first five flips. Event B is that we flip all heads on the second five flips."
#mutually exclusive
event3= "We roll a dice once. Event A is rolling an odd number. Event B is rolling a number greater than four."
#notmutuallyexclusive
event4 = "We have a bag of five marbles: three are green and two are blue. Suppose that we pick one marble from the bag. Event A is that the marble is green. Event B is that the marble is blue."
#not dependence
def guess_the_type(event):
    print(event)
    print(
        "Guess whether this event is a \n 1)dependent event \n 2)independent event \n 3)mutually exclusive event \n 4)not mutually exclusive event \n"
    )
    answer = int(input("Enter your answer : "))
    if event == event1:
        if answer == 1:
            print("\n Excellent! \n \n \n")
            return 'correct'
        else:
            print("\n Oops! You missed it... \n \n \n")
            return 'wrong'
    if event == event2:
        if answer == 3:
            print("\n Excellent! \n \n \n")
            return 'correct'
        else:
            print("\n Oops! You missed it... \n \n \n")
            return 'wrong'
    if event == event3:
        if answer == 4:
            print("\n Excellent! \n \n \n")
            return 'correct'
        else:
            print("\n Oops! You missed it... \n \n \n")
            return 'wrong'
    if event == event4:
        if answer == 2:
            print("\n Excellent! \n \n \n")
            return 'correct'
        else:
            print("\n Oops! You missed it... \n \n \n")
            return 'wrong'
List=[event1,event2,event3,event4]
Occur=random.sample(List,4)
firstanswer=guess_the_type(Occur[0])
secondanswer=guess_the_type(Occur[1])
thirdanswer=guess_the_type(Occur[2])
fourthanswer=guess_the_type(Occur[3])
if firstanswer=='correct' and secondanswer=='correct' and thirdanswer=='correct' and fourthanswer=='correct':
      print("\n \n \n Awesome! You have nailed all of them! Here's some cookies: 🍪🍪🍪 \n \n \n")
```
