```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
july_data = {'Day':['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
             'New Birth':[12,15,11,9,1,9,21]}
july_df = pd.DataFrame(july_data)
```

```python
plt.plot(july_df['Day'], july_df['New Birth'],
         marker='o',linestyle='--',
         linewidth=3,color='purple',label='July')
plt.title("New Birth Data-July")
plt.xlabel("Day of July")
plt.ylabel("New Birth")
plt.legend()
plt.show()
```

```python
aug_data = {'Day':['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
            'New Birth':[17,5,2,11,1,8,29]}
aug_df = pd.DataFrame(aug_data)
```

```python
plt.plot(aug_df['Day'], aug_df['New Birth'],
         marker='s',linestyle='-',
         linewidth=3,color='cyan',label='August')
plt.title("New Birth Data-August")
plt.xlabel("Day of August")
plt.ylabel("New Birth")
plt.legend()
plt.show()
```

```python
plt.plot(july_df['Day'], july_df['New Birth'],
         marker='o',linestyle='-',
         linewidth=3,color='purple',label='July')
plt.plot(aug_df['Day'], aug_df['New Birth'],
         marker='o',linestyle='-',
         linewidth=3,color='cyan',label='August')
plt.title("Comparison of new birth in 2 months")
plt.xlabel("Day of month")
plt.ylabel("New Birth")
plt.legend()
plt.show()
```

```python
x = np.arange(1,11,1)
y = 6*(x**2) + x + 1

plt.plot(x, y, marker='h', linestyle='-', linewidth=2, color='red')
plt.title("Equation y = 6x² + x + 1")
plt.xlabel("x values")
plt.ylabel("y values")
plt.show()
```
