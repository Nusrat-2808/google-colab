```
import sqlite3
```

```
conn=sqlite3.connect("people.db")
c=conn.cursor()
c.execute("CREATE TABLE IF NOT EXISTS PEOPLE(name TEXT,age INTEGER,profession TEXT)")
c.executemany("INSERT INTO PEOPLE VALUES(?,?,?)",
              [("sayeema",14,"student"),
               ("humayra",25,"teacher")])
for row in c.execute("SELECT * FROM PEOPLE"):
  print(row)
conn.commit()
conn.close()
```

```
#Connect with sqlite database
#Import necessary libraries
import sqlite3
database= '/content/database.sqlite'
conn=sqlite3.connect(database)
print('Opened database successfully')
```

```
#Read SQL query for getting all the tables of database into a dataframe
#Here SELECT* means select all
import pandas as pd
tables=pd.read_sql("""SELECT *
                  FROM sqlite_master
                  WHERE type='table';""",conn)
tables
```
