```python
df_cat=['gender','race/ethnicity','parental level of education','lunch','test preparation course']
for label in df_cat:
    sns.countplot(x=label,palette='mako',data=df)
    plt.show()
```

```python
df['race/ethnicity'].value_counts().head()
```

```python
race_cat=['group C','group D','group B','group E','group A']
plt.pie(df['race/ethnicity'].value_counts(),labels=race_cat)
```

```python
df['parental level of education'].value_counts().head(10)
```

```python
parent_edu_cat=['some college',"associate's degree",'high school','some high school',"bachelor's degree","master's degree"]
plt.pie(df['parental level of education'].value_counts(),labels=parent_edu_cat)
```

```python
sns.countplot(x='gender',hue='race/ethnicity',data=df,palette='pastel')
```

```python
sns.countplot(x='gender',hue='lunch',data=df,palette='pastel')
```

```python
for label in ['math score','reading score','writing score']:
    sns.distplot(df[label],kde=True)
    plt.show()
    print(f"Skewness of {label} : {df[label].skew()}")
```

```python
sns.boxplot(x='gender',y='math score',data=df)
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap='coolwarm')
plt.show()
```
