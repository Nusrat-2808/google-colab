```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Penguins Data.csv')
```

```python
plt.hist(df['Culmen Length (mm)'])
```

```python
plt.hist(df['Culmen Depth (mm)'])
```

```python
plt.hist(df['Flipper Length (mm)'])
```

```python
plt.hist(df['Body Mass (g)'])
```

```python
plt.hist('Culmen Length (mm)',data=df,width=0.5,color='C8')
```

```python
plt.hist('Culmen Depth (mm)',data=df,width=1,color='C1')
```

```python
plt.hist('Flipper Length (mm)',data=df,width=1.5,color='C2')
```

```python
plt.hist('Body Mass (g)',data=df,width=15,color='C3')
```
