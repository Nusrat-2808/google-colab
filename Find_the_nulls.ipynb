```python
import sqlite3
database='/content/basketball.sqlite'
conn=sqlite3.connect(database)
print("Opened data successfully")
```

```python
import pandas as pd
tables=pd.read_sql("""SELECT *
                        FROM sqlite_master
                        WHERE type='table';""",conn)
tables
```

```python
draft=pd.read_sql("""SELECT *
                        FROM Draft;""",conn)
draft
```

```python
null_Org=pd.read_sql("""SELECT *
                        FROM Draft
                        WHERE nameOrganizationFrom IS NULL;""",conn)
null_Org
```

```python
team=pd.read_sql("""SELECT *
                        FROM Team;""",conn)
team
```

```python
result1=pd.read_sql("""SELECT *
                        FROM Team
                        ORDER BY year_founded;""",conn)
result1.head()
```

```python
result1=pd.read_sql("""SELECT *
                        FROM Team
                        ORDER BY year_founded DESC;""",conn)
result1.head()
```
