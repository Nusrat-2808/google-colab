```python
df['Genre'].value_counts()
```

```python
genre_cat=['Non Fiction','Fiction']
df['Genre']=pd.Categorical(df['Genre'],genre_cat,ordered=True)
```

```python
median_index=np.median(df['Genre'].cat.codes)
median_genre=genre_cat[int(median_index)]
print(median_genre)
```

```python
plt.bar(genre_cat,df['Genre'].value_counts())
```

```python
plt.pie(df['Genre'].value_counts(),labels=genre_cat,colors=['pink','purple'])
```

```python
data=df.drop(['Name','Author','Genre'],axis=1)
```

```python
labels=['User Rating','Reviews','Price','Year']
```

```python
for label in labels:
  plt.boxplot(data[label])
  plt.title(label)
  plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
data=scaler.fit_transform(data)
```
